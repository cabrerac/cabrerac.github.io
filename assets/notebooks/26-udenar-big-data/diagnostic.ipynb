```
COLUMNS = [
    "age", "workclass", "fnlwgt", "education", "education-num",
    "marital-status", "occupation", "relationship", "race", "sex",
    "capital-gain", "capital-loss", "hours-per-week", "native-country", "income",
]
UCI_URL = "https://archive.ics.uci.edu/ml/machine-learning-databases/adult/adult.data"
COURSE_URL = "https://cabrerac.github.io/assets/data/26-udenar-big-data/adult.csv"
```

Ahora creamos una función para descargar y acceder al conjunto de datos.

```
def load_adult():
    # 1) UCI directo (canónico)
    try:
        df = pd.read_csv(UCI_URL, names=COLUMNS, skipinitialspace=True, na_values="?")
        print(f"Cargadas {df.shape[0]:,} filas desde UCI directo.")
        return df
    except Exception as e:
        print(f"UCI directo falló ({type(e).__name__}); probando paquete del curso...")
    # 2) Copia del curso en el sitio del instructor
    try:
        df = pd.read_csv(COURSE_URL)
        print(f"Cargadas {df.shape[0]:,} filas desde el paquete del curso.")
        return df
    except Exception as e:
        print(f"Paquete del curso falló ({type(e).__name__}); probando sklearn fetch_openml...")
    # 3) sklearn fetch_openml (último respaldo; requiere scikit-learn)
    from sklearn.datasets import fetch_openml
    bunch = fetch_openml("adult", version=2, as_frame=True)
    df = bunch.frame.copy()
    df.columns = [c.replace(".", "-") for c in df.columns]
    if "class" in df.columns:
        df = df.rename(columns={"class": "income"})
    print(f"Cargadas {df.shape[0]:,} filas desde sklearn fetch_openml.")
    return df
df = load_adult()
print("Forma:", df.shape)
df.head()
```

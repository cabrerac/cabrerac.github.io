# Practical Session 5: Regression and Classification Models

This lecture explores the regression and classification learning models. These models are simple but powerful and applicable in several domains. We'll define some crucial concepts of the ML process and then show the type of tasks where these models work. We'll also study the mathematical framework and the learning algorithms to train these models.

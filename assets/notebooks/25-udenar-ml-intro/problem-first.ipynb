# Practical Session 2: The Problem First

This lecture will emphasis on the importance of building ML-based systems with a purpose by focusing on the problem first. We will see the current status of ML applications, the adoption properties, and engineering mechanisms to ensure our ML projects align with the problems they are designed for.
